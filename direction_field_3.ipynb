```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Direction field for dy/dt = t - y
# ---------------------------------------------------

# Grid
t = np.linspace(0, 6, 25)
y = np.linspace(-3, 6, 25)

T, Y = np.meshgrid(t, y)

# Differential equation
M = T - Y

# Normalize vectors
L = np.sqrt(1 + M**2)

U = 1 / L
V = M / L

# ---------------------------------------------------
# Figure setup
# ---------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7, 7),
    constrained_layout=True
)

# ---------------------------------------------------
# Direction field
# ---------------------------------------------------

ax.quiver(
    T, Y,
    U, V,
    color='#00d9ff',
    angles='xy',
    scale_units='xy',
    scale=5.8,
    width=0.0032,
    pivot='middle',
    headwidth=3,
    headlength=4,
    headaxislength=3.5,
    alpha=0.9
)

# ---------------------------------------------------
# Particular solution
# y_p(t) = t - 1
# ---------------------------------------------------

t_curve = np.linspace(0, 6, 500)

y_p = t_curve - 1

ax.plot(
    t_curve,
    y_p,
    color='lime',
    linewidth=1.5,
    alpha=0.9,
    label=r"Particular solution $y_p(t)=t-1$"
)

# ---------------------------------------------------
# Nearby solution curves
# y(t) = t - 1 + c e^{-t}
# ---------------------------------------------------

constants = [-3, 2]

for i, c in enumerate(constants):

    y_curve = t_curve - 1 + c * np.exp(-t_curve)

    ax.plot(
        t_curve,
        y_curve,
        color='m',
        linewidth=1.5,
        alpha=0.75,
        label="Nearby solution curves" if i == 0 else None
    )

# ---------------------------------------------------
# Axes
# ---------------------------------------------------

ax.axhline(0, color='black', linewidth=1.1, alpha=0.55)
ax.axvline(0, color='black', linewidth=1.1, alpha=0.55)

# ---------------------------------------------------
# Labels and title
# ---------------------------------------------------

ax.set_xlabel(r"$t$", fontsize=13)
ax.set_ylabel(r"$y$", fontsize=13)

ax.set_title(
    r"Direction Field for $ y' = t-y$",
    fontsize=15,
    pad=14
)

# ---------------------------------------------------
# Limits
# ---------------------------------------------------

ax.set_xlim(0, 6)
ax.set_ylim(-3, 6)

# Equal aspect ratio makes one unit in t visually equal to one unit in y.
ax.set_aspect('equal', adjustable='box')

# ---------------------------------------------------
# Grid
# ---------------------------------------------------

ax.grid(
    True,
    alpha=0.25,
    linewidth=0.8
)

# ---------------------------------------------------
# Legend
# ---------------------------------------------------

ax.legend(
    fontsize=10,
    loc='upper left'
)

plt.show()
```
