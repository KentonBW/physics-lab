```python
import numpy as np
import matplotlib.pyplot as plt

# Domain
L = 1
x = np.linspace(0, L, 1000)

# Colors
wave_color = "#00d9ff"   # cyan
axis_color = "black"

# Create figure
fig, axes = plt.subplots(
    1, 4,
    figsize=(18, 3),
    facecolor="white"
)

for n, ax in enumerate(axes, start=1):

    # Stationary state
    psi = np.sin(n * np.pi * x / L)

    # Plot wavefunction
    ax.plot(
        x,
        psi,
        color=wave_color,
        linewidth=3
    )

    # Remove default frame
    for spine in ax.spines.values():
        spine.set_visible(False)

    # Remove ticks
    ax.set_xticks([])
    ax.set_yticks([])

    # Limits
    ax.set_xlim(-0.1, 1.15)
    ax.set_ylim(-1.8, 1.8)

    # Draw axes
    ax.plot(
        [0, 1.1],
        [0, 0],
        color=axis_color,
        linewidth=2
    )

    ax.plot(
        [0, 0],
        [-1.5, 1.5],
        color=axis_color,
        linewidth=2
    )

    # Arrowheads
    ax.plot(
        1.1, 0,
        marker=">",
        color=axis_color,
        markersize=10
    )

    ax.plot(
        0, 1.5,
        marker="^",
        color=axis_color,
        markersize=10
    )

    # Boundary mark at x = L
    ax.plot(
        [L, L],
        [-0.06, 0.06],
        color=axis_color,
        linewidth=2
    )

    # Labels
    ax.text(
        -0.06,
        1.42,
        rf"$\psi_{n}(x)$",
        color=axis_color,
        fontsize=20,
        ha="right"
    )

    ax.text(
        1.0,
        -0.48,
        r"$L$",
        color=axis_color,
        fontsize=18,
        ha="center"
    )

    ax.text(
        1.13,
        -0.03,
        r"$x$",
        color=axis_color,
        fontsize=18
    )

# Layout spacing
plt.tight_layout(w_pad=2)

plt.show()
```
