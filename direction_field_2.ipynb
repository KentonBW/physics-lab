```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Direction field for dy/dt = -y
# with solution curves
# ---------------------------------------------------

a = 1

# ---------------------------------------------------
# Grid
# ---------------------------------------------------

t = np.linspace(0, 5, 25)
y = np.linspace(-3, 3, 25)

T, Y = np.meshgrid(t, y)

# Differential equation
M = -a * Y

# Normalize vectors
L = np.sqrt(1 + M**2)

U = 1 / L
V = M / L

# ---------------------------------------------------
# Figure setup
# ---------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7, 5),
    constrained_layout=True
)

# ---------------------------------------------------
# Direction field
# ---------------------------------------------------

ax.quiver(
    T, Y,
    U, V,
    color="#00d9ff",
    angles='xy',
    scale_units='xy',
    scale=8,
    width=0.003,
    pivot='middle',
    headwidth=3,
    headlength=4,
    headaxislength=3.5,
    alpha=0.9
)

# ---------------------------------------------------
# Solution curves
# y(t) = c e^(-t)
# ---------------------------------------------------

t_sol = np.linspace(0, 5, 500)

initial_conditions = [-2, -1, 1, 2]

for i, y0 in enumerate(initial_conditions):

    y_sol = y0 * np.exp(-a * t_sol)

    ax.plot(
        t_sol,
        y_sol,
        color='m',
        linewidth=1.5,
        alpha=0.8,
        label="Solution curves" if i == 0 else None
    )

# Equilibrium solution
ax.plot(
    t_sol,
    np.zeros_like(t_sol),
    color='g',
    linewidth=1.5,
    linestyle='--',
    alpha=0.7,
    label=r"Equilibrium solution $y=0$"
)

# ---------------------------------------------------
# Axes
# ---------------------------------------------------

ax.axhline(
    0,
    color='black',
    linewidth=1.1,
    alpha=0.55
)

ax.axvline(
    0,
    color='black',
    linewidth=1.1,
    alpha=0.55
)

# ---------------------------------------------------
# Labels and title
# ---------------------------------------------------

ax.set_xlabel(
    r"$t$",
    fontsize=13
)

ax.set_ylabel(
    r"$y$",
    fontsize=13
)

ax.set_title(
    r"Direction Field for $ y' = -y$",
    fontsize=15,
    pad=14
)

# ---------------------------------------------------
# Limits
# ---------------------------------------------------

ax.set_xlim(0, 5)
ax.set_ylim(-3, 3)

# ---------------------------------------------------
# Grid
# ---------------------------------------------------

ax.grid(
    True,
    alpha=0.25,
    linewidth=0.8
)

# ---------------------------------------------------
# Legend
# ---------------------------------------------------

ax.legend(
    fontsize=10
)

plt.show()
```
