```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Direction field for dy/dt = y^2
# ---------------------------------------------------

# Grid
t = np.linspace(0, 4, 25)
y = np.linspace(-2.2, 2.2, 25)

T, Y = np.meshgrid(t, y)

# Differential equation
M = Y**2

# Use true geometric normalization
L = np.sqrt(1 + M**2)

U = 1 / L
V = M / L

# ---------------------------------------------------
# Figure setup
# ---------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7, 7),
    constrained_layout=True
)

# ---------------------------------------------------
# Direction field
# ---------------------------------------------------

ax.quiver(
    T, Y,
    U, V,
    color='#00d9ff',
    angles='xy',
    scale_units='xy',
    scale=6.5,
    width=0.003,
    pivot='middle',
    headwidth=3,
    headlength=4,
    headaxislength=3.5,
    alpha=0.9
)

# ---------------------------------------------------
# Smooth t values for solution curves
# ---------------------------------------------------

t_curve = np.linspace(0, 4, 1000)

# ---------------------------------------------------
# Solution curves
# y(t) = y0 / (1 - y0 t)
# ---------------------------------------------------

initial_values = [-2, -1, -0.5, 0.5, 0.7, 1]

for i, y0 in enumerate(initial_values):

    # Positive solutions blow up in finite time
    if y0 > 0:

        t_max = 0.98 / y0

        t_vals = np.linspace(
            0,
            min(4, t_max),
            700
        )

    else:

        t_vals = t_curve

    y_vals = y0 / (1 - y0 * t_vals)

    ax.plot(
        t_vals,
        y_vals,
        color='magenta',
        linewidth=1.5,
        alpha=0.75,
        label="Solution curves" if i == 0 else None
    )

# ---------------------------------------------------
# Equilibrium solution y = 0
# ---------------------------------------------------

ax.axhline(
    0,
    color='lime',
    linewidth=1.5,
    alpha=0.9,
    label=r"Equilibrium solution $y=0$"
)

# ---------------------------------------------------
# Axes
# ---------------------------------------------------

ax.axvline(
    0,
    color='black',
    linewidth=1.1,
    alpha=0.55
)

# ---------------------------------------------------
# Labels and title
# ---------------------------------------------------

ax.set_xlabel(r"$t$", fontsize=13)
ax.set_ylabel(r"$y$", fontsize=13)

ax.set_title(
    r"Direction Field for $ y' = y^2$",
    fontsize=15,
    pad=14
)

# ---------------------------------------------------
# Limits
# ---------------------------------------------------

ax.set_xlim(0, 4)
ax.set_ylim(-2.2, 2.2)

# ---------------------------------------------------
# Equal aspect ratio
# ---------------------------------------------------

ax.set_aspect('equal', adjustable='box')

# ---------------------------------------------------
# Grid
# ---------------------------------------------------

ax.grid(
    True,
    alpha=0.25,
    linewidth=0.8
)

# ---------------------------------------------------
# Legend
# ---------------------------------------------------

ax.legend(
    fontsize=10,
    loc='upper left'
)

plt.show()
```
