```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Resonant particular solution for
# y'' + 5y' + 4y = e^{-4t}
# ---------------------------------------------------

# Time values
t = np.linspace(0, 4, 1000)

# ---------------------------------------------------
# Resonant particular solution
# ---------------------------------------------------

y_p = -(1/3) * t * np.exp(-4*t)

# Forcing term for comparison
forcing = np.exp(-4*t)

# ---------------------------------------------------
# Figure setup
# ---------------------------------------------------

fig, ax = plt.subplots(
    figsize = (7, 5),
    constrained_layout = True
)

# ---------------------------------------------------
# Plot resonant particular solution
# ---------------------------------------------------

ax.plot(
    t,
    y_p,
    color = 'lime',
    linewidth = 1.5,
    alpha = 0.95,
    label = r"$y_p(t)=-\frac{1}{3}te^{-4t}$"
)

# ---------------------------------------------------
# Plot forcing term
# Scaled only for visual comparison
# ---------------------------------------------------

ax.plot(
    t,
    -0.08 * forcing,
    color = 'm',
    linewidth = 1.5,
    alpha = 0.8,
    label = r"Scaled $-e^{-4t}$"
)

# ---------------------------------------------------
# Axes
# ---------------------------------------------------

ax.axhline(0, color = 'black', linewidth = 1.1, alpha = 0.55)
ax.axvline(0, color = 'black', linewidth = 1.1, alpha = 0.55)

# ---------------------------------------------------
# Limits
# ---------------------------------------------------

ax.set_xlim(0, 4)
ax.set_ylim(-0.04, 0)

# ---------------------------------------------------
# Labels and title
# ---------------------------------------------------

ax.set_xlabel(r"$t$", fontsize = 13)
ax.set_ylabel(r"$y$", fontsize = 13)

ax.set_title(
    r"Resonant Particular Solution for $y''+5y'+4y=e^{-4t}$",
    fontsize = 15,
    pad = 14
)

# ---------------------------------------------------
# Legend
# ---------------------------------------------------

ax.legend(
    fontsize = 10,
    loc = 'lower right'
)

plt.show()
```
