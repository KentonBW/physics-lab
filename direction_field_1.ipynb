```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Direction field for dy/dx = x - y
# ---------------------------------------------------

# Differential equation
def f(x, y):
    return x - y

# Create grid
x = np.linspace(-3, 3, 25)
y = np.linspace(-3, 3, 25)

X, Y = np.meshgrid(x, y)

# Compute slopes
M = f(X, Y)

# Normalize vectors so all arrows have equal length
U = 1 / np.sqrt(1 + M**2)
V = M / np.sqrt(1 + M**2)

# ---------------------------------------------------
# Create figure
# ---------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7, 7),
    constrained_layout=True
)

# Direction field
ax.quiver(
    X, Y,
    U, V,
    angles='xy',
    pivot='middle',
    color='magenta',
    alpha=0.9
)

# Axes styling
ax.set_xlabel("x", fontsize=13)
ax.set_ylabel("y", fontsize=13)

ax.set_title(
    r"Direction Field for $ y' = x - y $",
    fontsize=15,
    pad=15
)

# Grid
ax.grid(
    True,
    alpha=0.25,
    linewidth=0.8
)

# Equal scaling
ax.set_aspect('equal')

plt.show()
```
